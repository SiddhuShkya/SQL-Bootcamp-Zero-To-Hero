```python
from sqlalchemy import create_engine, text
import pandas as pd

database = "dvdrental"
engine = create_engine(f"postgresql://postgres:shakya2830@localhost:5432/{database}")

# ✅ Wrap query with text()
query = text("""
SELECT 
  SUM(
  CASE rental_rate
    WHEN 0.99 THEN 1
    ELSE 0 
  END
  ) AS bargains,
  SUM(
  CASE rental_rate
    WHEN 2.99 THEN 1
    ELSE 0 
  END
  ) AS regular,
  SUM(
  CASE rental_rate
    WHEN 4.99 THEN 1
    ELSE 0 
  END
  ) AS premium
FROM film;
""")

df = pd.read_sql(query, engine)
df.head(5)
```
