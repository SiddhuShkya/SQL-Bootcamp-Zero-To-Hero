```python
from sqlalchemy import create_engine, text
import pandas as pd

database = "dvdrental"
engine = create_engine(f"postgresql://postgres:shakya2830@localhost:5432/{database}")

# ✅ Wrap query with text()
query = text("""
SELECT 
	SUM(
		CASE rating
		WHEN 'R' THEN 1
		ELSE 0
		END 
	) AS r,
	SUM(
		CASE rating
		WHEN 'PG' THEN 1
		ELSE 0
		END
	) AS pg,
	SUM(
	 	CASE rating
		 WHEN 'PG-13' THEN 1
		 ELSE 0
		 END
	) AS pg13
FROM film;
""")

df = pd.read_sql(query, engine)
df.head(5)
```
