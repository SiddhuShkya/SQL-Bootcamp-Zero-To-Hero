```python
from sqlalchemy import create_engine, text
import pandas as pd

database = "exercises"
engine = create_engine(f"postgresql://postgres:shakya2830@localhost:5432/{database}")

# ✅ Wrap query with text()
query = text("""
SELECT b.starttime
FROM cd.members m
INNER JOIN cd.bookings b
	ON m.memid = b.memid
WHERE firstname = 'David' AND surname = 'Farrell';
""")

df = pd.read_sql(query, engine)
df.head(10)
```
