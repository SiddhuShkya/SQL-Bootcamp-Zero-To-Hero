```python
import pandas as pd
from pandasql import sqldf
```

```python
TABLE_NAME1 = "film_table"
DATA_PATH1 = "../data/" + TABLE_NAME1 + ".csv"

TABLE_NAME2 = "inventory_table"
DATA_PATH2 = "../data/" + TABLE_NAME2 + ".csv"
df1 = pd.read_csv(DATA_PATH1)
df2 = pd.read_csv(DATA_PATH2)
# Important: assign dataframe to variable matching table name
df_table1 = df1
df_table2 = df2
query = """
SELECT df_table1.film_id, 
       df_table1.title, 
       df_table2.inventory_id, 
       df_table2.store_id
FROM df_table1
LEFT JOIN df_table2 
ON df_table2.film_id = df_table1.film_id
WHERE df_table2.film_id IS NULL
"""

result = sqldf(query, globals())
result.head(5)
```
